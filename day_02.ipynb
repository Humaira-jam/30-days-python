```python
firstname = 'Humaira'
lastname = 'Jamadar'
fullname = 'Humaira Salim Jamadar'
country = 'India'
city = 'Kurlap'
print (firstname , lastname , fullname, country , city)
print(len(firstname) , " is the length of first name")
print(len(lastname),len(fullname),len(country),len(city),
      " these are respective lengths of lastname, full name, country and city.")
num1 = 5
num2 = 4
total = float(num1+num2)
dif= float(num1-num2)
prod= float(num1*num2)
div=float(num1/num2)
rem= float(num1%num2)
exp=float(num2^num1)
floor_div= float(num1//num2)
print("Numbers 5 and 4...",
      "\naddition = ",total,
      "\ndiffference = ",dif,
      "\nproduct = ", prod,
      "\ndivision = ", div,
      "\nremainder = ", rem,
      "\nexponent funtion'4 to the power of 5' = ", exp,
      "\nfloor division = ", floor_div)
area_of_circle = 30*30*3.14
print("Area of circle (radius =30) = ",float(area_of_circle),"sq.m")
circum_of_circle = 2*30*3.14
print("circumference of same = ",float(circum_of_circle),"m")
```
