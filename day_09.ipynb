```python
marks=input('Enter your marks: ')
if float(marks)<=100 and float(marks)>=90:
  print('Grade: A')
elif float(marks)<=89 and float(marks)>=80:
  print('Grade: B')
elif float(marks)<=79 and float(marks)>=70:
  print('Grade: C')
elif float(marks)<=69 and float(marks)>=60:
  print('Grade: D')
else:
  print('Grade: F')
```

```python
season=input('Enter a month: ').lower()
if season == 'december' or 'january' or 'february':
  print("Winter season!!")
elif season == 'march' or 'april' or 'may' or 'june':
  print("Summer season!!")
elif season == 'july' or 'august' or 'septenber':
  print("Monsoon season!!")
else:
  print('Post monsoon!!')
```

```python
fruits = ['banana', 'orange', 'mango', 'lemon']
infruit = input('Enter a fruit:')
if infruit in fruits:
  print('Fruit already exists in the list!')
  print(fruits)
else:
  fruits.append(infruit)
  print(fruits)
```

```python
person={
    'first_name': 'Asabeneh',
    'last_name': 'Yetayeh',
    'age': 250,
    'country': 'Finland',
    'is_married': True,
    'skills': ['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
               }
       }
if 'skills' in person:
  mid_index_skills= int(len(person['skills'])/2)
  mid_skills = person.get('skills')[mid_index_skills]

  print('Skills: ', mid_skills)
```

```python
name_person = str(person['first_name'])+" "+str(person['last_name'])
if 'skills' in person:
  if 'Python' in person['skills']:
    print(f'{name_person} has Python as one of the skills.')
  else:
    print(f'{name_person} does not have Python as one of the skills.')
else:
  print(f'{name_person} does not have skills in their data.')
```

```python
if person['skills']==['Javascript','React']:
  print(f'{name_person} is a front-end developer')
elif person['skills']==['Node', 'Python', 'MongoDB']:
  print(f'{name_person} is a back-end developer')
elif person['skills']==['React', 'Node', 'MongoDB']:
  print(f'{name_person} is a full-stack developer')
elif 'Javascript' and 'React' and 'Node' and 'Python' and 'MongoDB' in person['skills']:
  print(f'{name_person} is an all-rounder developer')
```
