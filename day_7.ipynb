```python
#level_1
# sets
it_companies = {'Facebook', 'Google', 'Microsoft', 'Apple', 'IBM', 'Oracle', 'Amazon'}
print("No. of IT companies: ",len(it_companies))
it_companies.add('Twitter')
it_companies.update(['Meta', 'TCS', 'Wipro'])
it_companies.pop()
print(it_companies)
```

```python
#level_2
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
C=A.union(B)
print(C)
intersec=A.intersection(B)
print(intersec)
print("Is A subset of B: ",A.issubset(B))
print("Are A and B disjoint: ",A.isdisjoint(B))
A_new=A.union(B)
B_new=B.union(A)
print(A_new,B_new)
print("Symmetric Differenc: ",A.symmetric_difference(B))
del A
del B
```
