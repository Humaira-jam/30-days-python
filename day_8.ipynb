#Only one level : DICTIONARIES

```python
dog={'Name':'Tuffy',
     'Color':'White',
     'Breed':'Indian Spitz',
     'legs':4}
student={'first_name':'Humaira',
         'last_name':'Jamadar',
         'gender':'Female',
         'age':18,
         'marital status':'Unmarried',
         'skills':['Python Basics', 'swimming', 'sketching'],
         'country':'India',
         'address':{'city':'Islampur',
                    'district':'Sanglali',
                    'taluka':'Walwa',
                    'landmark':'Near Gram Panchayat, Kurlap'
                    }
         }
print("elements in student dictionary: ", len(student))
print("Student's Skill set: ",student.get('skills'),'\n',type(student['skills']))
extra_skills=input("Enter any more skills: ").split(',')
student['skills'].extend(extra_skills)
print(student["skills"])
attributes=student.keys()
print(attributes)
info=student.values()
print(info)
print(student.items())
del dog['legs']
del dog
```
